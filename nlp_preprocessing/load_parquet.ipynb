```python
#FILTER ONLY HTTP-LIKE PAYLOADS:
def is_http(payload):
    if not isinstance(payload, str): return False
    return payload.lower().startswith(("get ", "post ", "head ")) and "http/1.1" in payload.lower()

df = df_parquet[df_parquet["PayloadStrings"].apply(is_http)].copy()
df.head(10)
```

```python
import string
```

```python
encoding_types=df_parquet['PayloadEncoding'].unique().tolist()
print(encoding_types)
```

```python
def decode_payload(payload):
    if not isinstance(payload,str):
        return ""
    byte_payload= payload.encode('latin-1', errors='ignore')
    for enc in encoding_types:
        try:
            decoded= byte_payload.decode(enc)
        #check if the decoded strings has enough readable and real-like characters to consider it valid, if not => remove them:
        #they are not meaningful for nlp and can trick the model , as it can contain corrupted data in a benign labelling.
            ascii_letters = string.ascii_letters + string.digits + " "
            num_clean = sum(1 for c in decoded if c in ascii_letters)
            if num_clean / max(len(decoded), 1) > 0.6:
                return decoded
        except :
            continue
    return ""

#check whether it's a payoload or a log per attack :
attack_types=df['Attack'].dropna().unique()
pd.set_option("display.max_colwidth", None)  
pd.set_option("display.max_rows", 1000)

for attack in attack_types:
    print(f"\n Attack type: {attack}")
    samples=(df[df['Attack']==attack].copy()
             .dropna(subset=["PayloadStrings"]).head(10))
    for i, raw in enumerate(samples["PayloadStrings"]):
        decoded_payload=decode_payload(raw)
        if decoded_payload:
            print(f"\n --Payload {i+1} --\n{decoded_payload.strip()[:1000]}")
```

```python
df['PayloadStrings'] = df['PayloadStrings'].apply(decode_payload)
df= df[df['PayloadStrings'].str.strip() != ""]
```

```python
def clean_payload_basic(payload):
    if not isinstance(payload, str):
        return ""
    payload = re.sub(r"[\x00-\x1f\x7f]", " ", payload)          
    payload = re.sub(r"[\[\]{}<>]", " ", payload)               
    payload = re.sub(r"\s+", " ", payload)                      
    return payload.strip()

df['PayloadStrings'] = df['PayloadStrings'].apply(clean_payload_basic)
#df['nlp_text'] = df['PayloadStrings'].apply(clean_text)
```

```python
def smart_split(payload):
    if not isinstance(payload, str):
        return payload
    if "\n" not in payload:
        payload = payload.replace(" HTTP/1.1 ", " HTTP/1.1\n")
        payload = re.sub(r"(Host:|User-Agent:|Referer:|Content-Type:|Cookie:)", r"\n\1", payload)
    return payload
df["PayloadStrings"] = df["PayloadStrings"].apply(smart_split)

```

```python
import re, json
from urllib.parse import urlparse, parse_qs

def extract_from_payload_string(payload):
    try:
        if not isinstance(payload, str):
            return None

        lines = [l.strip() for l in payload.split("\n") if l.strip()]
        if not lines or len(lines[0].split()) < 2:  # More general, just method and path
            return None

        #Safe extraction of request line
        parts = lines[0].split()
        method = parts[0]
        full_path = parts[1] if len(parts) > 1 else "/"
        parsed = urlparse(full_path)
        path = parsed.path
        query = parsed.query
        query_dict = parse_qs(query)

        # extract headers and body
        headers = {}
        body = ""
        is_body = False

        for line in lines[1:]:
            if ":" in line and not is_body:
                k, v = line.split(":", 1)
                headers[k.strip().lower()] = v.strip()
            else:
                is_body = True
                body += line + " "

        #combine all attack surfaces
        combined = " ".join([
            body, query,
            headers.get("cookie", ""),
            headers.get("user-agent", ""),
            headers.get("referer", "")
        ]).lower()

        #matched keywords
        sql_patterns = [
            r"\bselect\b", r"\binsert\b", r"\bupdate\b", r"\bdelete\b", r"\bdrop\b", r"\bunion\b",
            r"\b(and|or)\b\s+[^\s]+\s*=\s*[^\s]+",
            r"['\"]\s*or\s*['\"]?\d+=\d+",
            r"\b\d+\s*=\s*\d+\b"
        ]
        sql_keywords = []
        for pattern in sql_patterns:
            sql_keywords += re.findall(pattern, combined)

        xss_patterns = [
            r"<script.*?>.*?</script>", r"(on\w+\s*=)", r"(document\.cookie)", r"(alert\s*\()",
            r"(eval\s*\()", r"(<img\s+[^>]*onerror\s*=)", r"(<svg\s+[^>]*onload\s*=)"
        ]
        xss_keywords = []
        for pattern in xss_patterns:
            xss_keywords += re.findall(pattern, combined)

        shell_patterns = [
            r"(curl|wget|bash|sh|nc|netcat|python|perl|powershell|cmd|scp|ftp)",
            r"(cmd(\.exe)?|powershell(\.exe)?)",
            r"/bin/(sh|bash|nc|cat)",
            r"(bash|sh)\s+-i",
            r"nc\s+-e\s+/bin/sh",
            r"exec\s+/bin/sh",
            r"(bash|sh)\s+-c\s+['\"].+?['\"]",
            r"powershell\s+-[eE]ncode[d]?",
            r"(?:wget|curl)\s+http[s]?://",
            r"rm\s+-rf\s+/",
            r"scp\s+\S+\s+\S+",
            r"\$\((.*?)\)",
            r"`.*?`",
            r"base64\s+-d",
            r"(?:(?:curl|wget|bash|sh|nc|netcat|python|perl|powershell|cmd|scp|ftp)[^;|&]{0,40})(\||;|&&)"
        ]
        shell_keywords = []
        for pattern in shell_patterns:
            shell_keywords += re.findall(pattern, combined)

        return {
            "method": method,
            "url_path": path,
            "query_parameters": json.dumps(query_dict),
            "user_agent": headers.get("user-agent", ""),
            "referer": headers.get("referer", ""),
            "cookies": headers.get("cookie", ""),
            "body": body.strip(),
            "sql_keywords": sql_keywords,
            "xss_keywords": xss_keywords,
            "shell_keywords": shell_keywords,
            "is_malicious": bool(sql_keywords or xss_keywords or shell_keywords)
        }

    except Exception as e:
        return None
```
